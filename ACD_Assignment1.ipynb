# 1. DFA Simulator

A Deterministic Finite Automaton (DFA) has exactly one possible transition
for each state and input symbol.

The DFA used in this assignment accepts binary strings containing the
substring `01`.

Alphabet:

Σ = {0, 1}


```python
# ==========================================
# DFA SIMULATOR
# ==========================================

def dfa_simulator(states, alphabet, start_state,
                  accept_states, transitions, input_string):

    current_state = start_state

    print("DFA Simulation")
    print("-" * 30)
    print("Start State:", current_state)

    for symbol in input_string:

        if symbol not in alphabet:
            print("Invalid symbol:", symbol)
            return False

        current_state = transitions[current_state][symbol]

        print(f"Read '{symbol}' -> {current_state}")

    if current_state in accept_states:
        print("\nResult: ACCEPTED")
        return True
    else:
        print("\nResult: REJECTED")
        return False
```

```python
# ==========================================
# DFA DEFINITION
# ==========================================

states = {"A", "B", "C", "D"}
alphabet = {"0", "1"}

start_state = "A"
accept_states = {"C", "D"}

transitions = {
    "A": {
        "0": "B",
        "1": "A"
    },

    "B": {
        "0": "B",
        "1": "C"
    },

    "C": {
        "0": "D",
        "1": "C"
    },

    "D": {
        "0": "D",
        "1": "C"
    }
}

print("DFA created successfully.")
```

# 2. NFA Simulator

A Non-Deterministic Finite Automaton (NFA) can have zero, one, or multiple
possible transitions for a state and input symbol.

The NFA keeps track of a set of possible current states.

The NFA used in this assignment accepts binary strings containing `01`.


```python
# ==========================================
# NFA SIMULATOR
# ==========================================

def nfa_simulator(states, alphabet, start_state,
                  accept_states, transitions, input_string):

    current_states = {start_state}

    print("NFA Simulation")
    print("-" * 30)
    print("Initial States:", current_states)

    for symbol in input_string:

        if symbol not in alphabet:
            print("Invalid symbol:", symbol)
            return False

        next_states = set()

        for state in current_states:

            possible_states = transitions.get(
                state, {}
            ).get(symbol, set())

            next_states.update(possible_states)

        current_states = next_states

        print(f"Read '{symbol}' -> {current_states}")

    if current_states & accept_states:
        print("\nResult: ACCEPTED")
        return True
    else:
        print("\nResult: REJECTED")
        return False
```

```python
# ==========================================
# NFA DEFINITION
# ==========================================

states = {"q0", "q1", "q2"}
alphabet = {"0", "1"}

start_state = "q0"
accept_states = {"q2"}

transitions = {

    "q0": {
        "0": {"q0", "q1"},
        "1": {"q0"}
    },

    "q1": {
        "0": set(),
        "1": {"q2"}
    },

    "q2": {
        "0": {"q2"},
        "1": {"q2"}
    }
}

print("NFA created successfully.")
```

# 3. NFA to DFA Conversion

The NFA is converted into an equivalent DFA using the Subset Construction
method.

In this method, each DFA state represents a subset of NFA states.

For example:

{q0, q1}

is treated as one DFA state.

A DFA state is accepting if it contains at least one accepting state of
the original NFA.


```python
# ==========================================
# NFA TO DFA - SUBSET CONSTRUCTION
# ==========================================

from collections import deque


def nfa_to_dfa(states, alphabet, start_state,
               accept_states, transitions):

    start_subset = frozenset([start_state])

    queue = deque([start_subset])

    visited = {start_subset}

    dfa_transitions = {}

    while queue:

        current_subset = queue.popleft()

        dfa_transitions[current_subset] = {}

        for symbol in alphabet:

            next_states = set()

            for state in current_subset:

                possible_states = transitions.get(
                    state, {}
                ).get(symbol, set())

                next_states.update(possible_states)

            next_subset = frozenset(next_states)

            dfa_transitions[current_subset][symbol] = next_subset

            if next_subset not in visited:

                visited.add(next_subset)
                queue.append(next_subset)

    dfa_accept_states = {
        subset
        for subset in visited
        if subset & accept_states
    }

    return (
        visited,
        dfa_transitions,
        start_subset,
        dfa_accept_states
    )
```

```python
# ==========================================
# CONVERT NFA TO DFA
# ==========================================

dfa_states, dfa_transitions, dfa_start, dfa_accept = nfa_to_dfa(
    states,
    alphabet,
    start_state,
    accept_states,
    transitions
)

print("NFA successfully converted to DFA.")
```

```python
print("DFA States")
print("=" * 40)

for state in dfa_states:
    print(set(state))

print("\nDFA Start State")
print("=" * 40)

print(set(dfa_start))

print("\nDFA Accepting States")
print("=" * 40)

for state in dfa_accept:
    print(set(state))

print("\nDFA Transition Table")
print("=" * 40)

for state in dfa_transitions:

    for symbol in alphabet:

        destination = dfa_transitions[state][symbol]

        print(
            f"{set(state)} -- {symbol} --> {set(destination)}"
        )
```

## Testing the Converted DFA

The converted DFA should give the same acceptance result as the original
NFA for the same input strings.


```python
def simulate_converted_dfa(
    dfa_transitions,
    dfa_start,
    dfa_accept,
    input_string,
    alphabet
):

    current_state = dfa_start

    print("Converted DFA Simulation")
    print("-" * 35)

    print("Start:", set(current_state))

    for symbol in input_string:

        if symbol not in alphabet:
            print("Invalid symbol:", symbol)
            return False

        current_state = dfa_transitions[
            current_state
        ][symbol]

        print(
            f"Read '{symbol}' -> {set(current_state)}"
        )

    if current_state in dfa_accept:

        print("\nResult: ACCEPTED")
        return True

    else:

        print("\nResult: REJECTED")
        return False
```

```python
simulate_converted_dfa(
    dfa_transitions,
    dfa_start,
    dfa_accept,
    "001",
    alphabet
)
```

```python
simulate_converted_dfa(
    dfa_transitions,
    dfa_start,
    dfa_accept,
    "111",
    alphabet
)
```
